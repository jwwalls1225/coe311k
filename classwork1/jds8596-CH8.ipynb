```python
import numpy as np

def diagdom(A):
    n = len(A)
    for i in range(n):
        diag = abs(A[i][i])
        diagsum = sum(abs(A[i][j]) for j in range(n) if j != i)
        if diag < diagsum:
            return False
    return True

A = np.array([#insert example matrix here])
result = diagdom(A)

print("Is the matrix diagonally dominant?", result)
print(A)
print("Is the matrix diagonally dominant?", result)
```
