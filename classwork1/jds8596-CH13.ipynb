```python
import numpy as np
import time
import matplotlib.pyplot as plt

def naive(A, B):
    n = len(A)
    C = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

sizes = [3 * (2**i) for i in range(8)]
matrix_numbers = list(range(1, 9))

naive_times = []
numpy_times = []

for n in sizes:
    A = np.random.rand(n, n)
    B = np.random.rand(n, n)

    start = time.perf_counter()
    naive(A, B)
    naive_times.append(time.perf_counter() - start)

    start = time.perf_counter()
    np.dot(A, B)
    numpy_times.append(time.perf_counter() - start)

plt.plot(matrix_numbers, naive_times, marker='o')
plt.plot(matrix_numbers, numpy_times, marker='o')

plt.xlabel("Matrix Number (1–8)")
plt.ylabel("Runtime (seconds)")
plt.title("Naive vs NumPy Matrix Multiplication Runtime")
plt.legend(["Naive (Custom)", "NumPy (Optimized)"])
plt.grid(True)
plt.show()
```
