```python
import numpy as np
import matplotlib.pyplot as plt

def f(t, y):
    return -1000*y + 3000 - 2000*np.exp(-t)

def exact(t):
    return 3 - 0.998*np.exp(-1000*t) - 2.002*np.exp(-t)

h_values = [0.1, 0.01, 0.001, 0.0001, 0.00001]

t0, tf = 0, 30
y0 = 0

plt.figure()

for h in h_values:
    t = np.arange(t0, tf + h, h)
    y = np.zeros(len(t))
    y[0] = y0

    for i in range(len(t) - 1):
        y[i+1] = y[i] + h * f(t[i], y[i])

    plt.plot(t, y, label=f'h = {h}')

t_dense = np.linspace(t0, tf, 1000)
plt.plot(t_dense, exact(t_dense), 'k--', label='Exact')

plt.title("Euler Forward vs Analytical Solution")
plt.xlabel("t")
plt.ylabel("y(t)")
plt.legend()
plt.grid()
plt.show()
```
