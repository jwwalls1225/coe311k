```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5])
Y = np.array([1.2, 2.8, 3.6, 4.5, 5.1])

A = np.vstack([X, np.ones(len(X))]).T

m, c = np.linalg.lstsq(A, Y, rcond=None)[0]

print(f"Slope: {m}, Intercept: {c}")

plt.figure()
plt.scatter(X, Y)
plt.plot(X, m*X + c)
plt.show()

perm = np.random.permutation(len(X))
X_rand = X[perm]
Y_rand = Y[perm]

A_rand = np.vstack([X_rand, np.ones(len(X_rand))]).T
m_rand, c_rand = np.linalg.lstsq(A_rand, Y_rand, rcond=None)[0]

print(f"Slope: {m_rand}, Intercept: {c_rand}")


plt.figure()
plt.scatter(X_rand, Y_rand)
plt.plot(X_rand, m_rand*X_rand + c_rand)
plt.show()

x = np.arange(0, 1.01, 0.01)
y = 3*x - 0.5

perturbation = np.random.normal(0, 0.1, len(x))
y_pert = y + perturbation

A1 = np.vstack([x, np.ones(len(x))]).T
m1, c1 = np.linalg.lstsq(A1, y_pert, rcond=None)[0]

print("Original slope: 3")
print("Original intercept: -0.5")
print(f"New slope: {m1}")
print(f"New intercept: {c1}")

plt.figure()
plt.scatter(x, y_pert)
plt.plot(x, m1*x + c1)
plt.show()
```
