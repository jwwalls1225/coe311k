```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[0, 1], [-2, -0.5]])

x0 = np.array([1, 0])

t0 = 0
tf = 10

h_values = np.concatenate((np.arange(0.01, 0.26, 0.04),
                           np.arange(0.25, 1.05, 0.05)))

for h in h_values:
    E = np.eye(A.shape[0]) + h * A

    eigenvalues = np.linalg.eigvals(E)

    stable = np.all(np.abs(eigenvalues) < 1)

    print(f"\nStep size h = {h:.2f}")
    print("Eigenvalues:", eigenvalues)

    if stable:
        print("Solution is Stable")
    else:
        print("Solution may not be Entirely Stable ;)")

    t = np.arange(t0, tf + h, h)

    x = np.zeros((len(t), len(x0)))
    x[0] = x0

    for i in range(len(t) - 1):
        x[i + 1] = x[i] + h * A @ x[i]

    plt.plot(t, x[:, 0], label=f'h = {h:.2f}')

plt.title("Euler Method Solutions for Different Step Sizes")
plt.xlabel("Time")
plt.ylabel("x₁(t)")
plt.legend()
plt.grid(True)
plt.show()
```
