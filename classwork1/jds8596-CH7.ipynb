```python
import numpy as np
import time
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        if np.linalg.norm(A @ x_new - b, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_iterations

def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        if np.linalg.norm(A @ x_new - b, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_iterations


tol = 1e-6
max_iterations = 1000

A1 = np.array([
[0.582745, 0.48 , 0.10 , 0.  , 0.  ],
[0.48 , 1.044129, 0.46 , 0.10 , 0.  ],
[0.10 , 0.46 , 1.10431 , 0.44 , 0.10 ],
[0.  , 0.10 , 0.44 , 0.963889, 0.42 ],
[0.  , 0.  , 0.10 , 0.42 , 0.522565]], dtype=float)

b1 = np.array([1.162745, 2.084129, 2.20431 , 1.923889, 1.042565], dtype=float)

x0 = np.zeros_like(b1)

start = time.time()
gs_sol, gs_it = gauss_seidel(A1, b1, x0, tol, max_iterations)
print("Gauss-Seidel:", gs_sol)
print("Iterations:", gs_it)
print("Residual:", np.linalg.norm(A1 @ gs_sol - b1, ord=np.inf))
print("Time:", time.time() - start)

start = time.time()
jac_sol, jac_it = jacobi(A1, b1, x0, tol, max_iterations)
print("\nJacobi:", jac_sol)
print("Iterations:", jac_it)
print("Residual:", np.linalg.norm(A1 @ jac_sol - b1, ord=np.inf))
print("Time:", time.time() - start)

A2 = np.array([[1, 1],
               [1, 1.0001]], dtype=float)

b2 = np.array([1, 1], dtype=float)

x0 = np.zeros_like(b2)

gs_sol, gs_it = gauss_seidel(A2, b2, x0, tol, max_iterations)
print("Gauss-Seidel:", gs_sol)
print("Iterations:", gs_it)
print("Residual:", np.linalg.norm(A2 @ gs_sol - b2, ord=np.inf))

jac_sol, jac_it = jacobi(A2, b2, x0, tol, max_iterations)
print("\nJacobi:", jac_sol)
print("Iterations:", jac_it)
print("Residual:", np.linalg.norm(A2 @ jac_sol - b2, ord=np.inf))

A3 = np.array([[1, 1],
               [1, 1.0001]], dtype=float)

b3 = np.array([1, 1.0001], dtype=float)

x0 = np.zeros_like(b3)

gs_sol, gs_it = gauss_seidel(A3, b3, x0, tol, max_iterations)
print("Gauss-Seidel:", gs_sol)
print("Iterations:", gs_it)
print("Residual:", np.linalg.norm(A3 @ gs_sol - b3, ord=np.inf))

jac_sol, jac_it = jacobi(A3, b3, x0, tol, max_iterations)
print("\nJacobi:", jac_sol)
print("Iterations:", jac_it)
print("Residual:", np.linalg.norm(A3 @ jac_sol - b3, ord=np.inf))
```
