```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sin(x)

a = 0
b = 2 * np.pi
N = 50

x = np.linspace(a, b, N)

h = (b - a) / (N - 1)

true_derivative = np.cos(x)

forward_diff = (f(x + h) - f(x)) / h
backward_diff = (f(x) - f(x - h)) / h
central_diff = (f(x + h) - f(x - h)) / (2 * h)

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.plot(x, true_derivative)
plt.title("True Derivative: cos(x)")

plt.subplot(2, 2, 2)
plt.plot(x, true_derivative, label="cos(x)")
plt.plot(x, forward_diff, '--', label="Forward")
plt.title("Forward Difference vs True")
plt.legend()

plt.subplot(2, 2, 3)
plt.plot(x, true_derivative, label="cos(x)")
plt.plot(x, backward_diff, '--', label="Backward")
plt.title("Backward Difference vs True")
plt.legend()

plt.subplot(2, 2, 4)
plt.plot(x, true_derivative, label="cos(x)")
plt.plot(x, central_diff, '--', label="Central")
plt.title("Central Difference vs True")
plt.legend()

plt.tight_layout()
plt.show()
```
