```python
import numpy as np
v1 = np.array([1, 2, 3]) # 3-dimensional vector
v2 = np.array([4, 5, 6]) 
print(v1)
print(v2)
```

```python
v_add = v1 + v2
v_sub = v1 - v2
print(v_add)
print(v_sub)
```

```python
scalar = 2
v1_scaled = scalar * v1
print(v1_scaled)
```

```python
dot_product = np.dot(v1, v2)
print(dot_product)
```

```python
magnitude_v1 = np.linalg.norm(v1)
print(magnitude_v1)
# magnitude is the sqare root of the sum of the squares of its components
```

```python
A = np.array([[1, 2, 3], [4, 5, 6]])
print(A)
# 2x3 matrix

B = np.array([[7, 8, 9], [10, 11, 12], [13, 14, 15]])
print(B)
# 3x3 matrix
```

```python
C = np.array([[1, 2, 3], [4, 5, 6]])
matrix_add = A + C
matrix_sub = A - C
print("A+C= ", matrix_add)
print("A-C= ", matrix_sub)
```

```python
matrix_mul = np.dot(A, B)
print(matrix_mul)
```

```python
transpose_A = A.T
print(transpose_A)
```

```python
a = np.array([[2, 3], [4, -1]])
print(a)
b = np.array([5, 3])
print(b)
x = np.linalg.solve(a, b)
print(x)
```

```python
#Eigenvalues and Eigenvectors
#Define matrix A
A = np.array([[2, 1], [1, 2]])
#Compute eigens
e_values, e_vectors = np.linalg.eig(A)
print("Eigenvalues: ", e_values)
print("Eigenvectors: ", e_vectors)
```

```python
#Inverse
A = np.array([[4, 7], [2, 6]])
A_inv = np.linalg.inv(A)
print(A)
print(A_inv)
```

```python
identity = np.dot(A, A_inv)
print(identity)
```

```python
# Define matrix
A = np.array([[1, 2], [3, 4], [5, 6]])
# Perform Singular Value Decomposition (SVD)
U, sigma, Vt = np.linalg.svd(A)
print("U matrix:\n ", U)
print("Sigma values:\n ", sigma)
print("V^T matrix:\n ", Vt)
```
