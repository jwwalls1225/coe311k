We can do a Stability Analysis on -2y*cos(x). We can treat the equation as λ(x)= −2cos(x), applying the stability condition ∣1+hλ∣ < 1 gives us a local stability condition. And since λ(x) only has values from -2 to 2, the bound gives us 0 < h < 1. However, Euler's can't stay stable because when cos(x) < 0, λ(x) is positive, and the system grows.
