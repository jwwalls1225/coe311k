```python
import numpy as np
import math
import matplotlib.pyplot as plt

def taylor(x, terms):
    calc_value = 0
    for i in range(terms):
        new_value = ((-1)**i) * (x**(2*i + 1)) / math.factorial(2*i + 1)
        calc_value = calc_value + new_value
    return calc_value
xvals = np.arange(0, 10.1, 0.1)

exact = np.sin(xvals)

errors = {1: [], 2: [], 3: [], 4: []}
for n in [1, 2, 3, 4]:
    approx = np.array([taylor(x, n) for x in xvals])
    error = np.abs(approx - exact)
    errors[n] = error
    

for n in [1, 2, 3, 4]:
    plt.plot(xvals, errors[n], label=f"{n} terms") #Used Copilot to figure out how to label the legend
plt.legend()
plt.show()

```

```python
## Discussion: The error does not decrease uniformly for all values of x,
# because the further you move from the initial value, the less accurate
# the approximation is. The truncation error behaves differently for 
# small vs. values of x, in that the error is very high for the first few
# terms, but as you add terms, the error decreases very rapidly. 
```
