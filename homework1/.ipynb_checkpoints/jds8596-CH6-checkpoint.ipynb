```python
import numpy as np
import time
def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        if np.linalg.norm(A @ x_new - b, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_iterations
A = np.array([[10, -1, 2, 0, 0],
              [-1, 11, -1, 3, 0],
              [2, -1, 10, -1, 0],
              [0, 3, -1, 8, -2],
              [0, 0, 0, -2, 9]], dtype=float)
b = np.array([14, 30, 26, 25, 37], dtype=float)
x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100
start = time.time()
solution, iterations = gauss_seidel(A, b, x0, tol, max_iterations)
end = time.time()
print("Gauss solution: ", solution)
print("Gauss iterations: ", iterations)
print("Gauss time: ", end - start)
def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        if np.linalg.norm(A @ x_new - b, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_iterations
start = time.time()
solution, iterations = jacobi(A, b, x0, tol, max_iterations)
end = time.time()
print("Jacobi solution: ", solution)
print("Jacobi iterations: ", iterations)
print("Jacobi time: ", end - start)
print("Final residual:", np.linalg.norm(A @ solution - b, ord=np.inf))
```
